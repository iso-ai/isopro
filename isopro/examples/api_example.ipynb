# ISOPro Core Simulation API Example

This notebook demonstrates how to use the ISOPro Core Simulation API to interact with the various simulation types.

```python
import requests
import json
import os
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# API base URL - change this to your deployed API URL if using remote deployment
API_BASE_URL = "http://localhost:5000"
```

## Health Check

Let's start by checking if the API is running:

```python
response = requests.get(f"{API_BASE_URL}/healthcheck")
print(f"Status Code: {response.status_code}")
print(json.dumps(response.json(), indent=2))
```

## Reasoning Simulation Example

The `/simulate/reason` endpoint allows you to run reasoning simulations with a prompt and optional context:

```python
reason_data = {
    "prompt": "Explain the advantages and disadvantages of electric vehicles compared to traditional combustion engine vehicles.",
    "context": "Focus on environmental impact, cost factors, and infrastructure requirements.",
    "model": "claude-3-sonnet-20240229"
}

response = requests.post(f"{API_BASE_URL}/simulate/reason", json=reason_data)
print(f"Status Code: {response.status_code}")
result = response.json()

print(f"Run ID: {result['run_id']}")
print(f"\nOutput:\n{result['output']}")
print(f"\nMetadata:\n{json.dumps(result['metadata'], indent=2)}")
```

## Question-Answering Simulation Example

The `/simulate/qa` endpoint allows you to run Q&A simulations with different personas:

```python
qa_data = {
    "question": "Why is my internet connection so slow?",
    "persona_type": "upset",
    "num_turns": 2,
    "model": "claude-3-sonnet-20240229"
}

response = requests.post(f"{API_BASE_URL}/simulate/qa", json=qa_data)
print(f"Status Code: {response.status_code}")
result = response.json()

print(f"Run ID: {result['run_id']}")
print("\nConversation:")
for i, message in enumerate(result['output']):
    print(f"Turn {i+1}:")
    print(f"  User: {message.get('user_message', '')}")
    print(f"  AI: {message.get('ai_response', '')}")
    print("")
print(f"\nMetadata:\n{json.dumps(result['metadata'], indent=2)}")
```

## Custom Persona Example

You can also create custom personas for the Q&A simulation:

```python
custom_qa_data = {
    "question": "I need help with your product pricing.",
    "custom_persona": {
        "name": "Business Client",
        "characteristics": ["busy", "detail-oriented", "budget-conscious"],
        "message_templates": [
            "I need to understand your enterprise pricing structure immediately.",
            "What kind of volume discounts do you offer?"
        ]
    },
    "num_turns": 2,
    "model": "claude-3-sonnet-20240229"
}

response = requests.post(f"{API_BASE_URL}/simulate/qa", json=custom_qa_data)
print(f"Status Code: {response.status_code}")
result = response.json()

print(f"Run ID: {result['run_id']}")
print("\nConversation:")
for i, message in enumerate(result['output']):
    print(f"Turn {i+1}:")
    print(f"  User: {message.get('user_message', '')}")
    print(f"  AI: {message.get('ai_response', '')}")
    print("")
print(f"\nMetadata:\n{json.dumps(result['metadata'], indent=2)}")
```

## Adversarial Simulation Example

The `/simulate/adversarial` endpoint allows you to test model robustness against various attack types:

```python
adversarial_data = {
    "prompt": "What is the capital of France?",
    "attack_type": "prompt_injection",
    "num_steps": 1,
    "model": "claude-3-sonnet-20240229"
}

response = requests.post(f"{API_BASE_URL}/simulate/adversarial", json=adversarial_data)
print(f"Status Code: {response.status_code}")
result = response.json()

print(f"Run ID: {result['run_id']}")
print("\nAdversarial Results:")
for i, test in enumerate(result['output']):
    print(f"Test {i+1}:")
    print(f"  Original Input: {test.get('original_input', '')}")
    print(f"  Perturbed Input: {test.get('perturbed_input', '')}")
    print(f"  Original Output: {test.get('original_output', '')}")
    print(f"  Perturbed Output: {test.get('perturbed_output', '')}")
    print("")
print(f"\nMetadata:\n{json.dumps(result['metadata'], indent=2)}")
```

## Orchestration Simulation Example

The `/simulate/orchestration` endpoint allows you to run multi-agent orchestration simulations:

```python
orchestration_data = {
    "input_data": "Analyze the current trends in renewable energy adoption worldwide and predict the most promising technologies for the next decade.",
    "mode": "agent"
}

response = requests.post(f"{API_BASE_URL}/simulate/orchestration", json=orchestration_data)
print(f"Status Code: {response.status_code}")
result = response.json()

print(f"Run ID: {result['run_id']}")
print(f"\nOrchestration Results:\n{json.dumps(result['output'], indent=2)}")
print(f"\nMetadata:\n{json.dumps(result['metadata'], indent=2)}")
```

## Using the Base /simulate Endpoint

You can also use the base `/simulate` endpoint with a 'type' parameter to route to the appropriate simulation:

```python
base_sim_data = {
    "type": "reason",
    "prompt": "Explain how neural networks learn patterns in data.",
    "model": "claude-3-sonnet-20240229"
}

response = requests.post(f"{API_BASE_URL}/simulate", json=base_sim_data)
print(f"Status Code: {response.status_code}")
result = response.json()

print(f"Run ID: {result['run_id']}")
print(f"\nOutput:\n{result['output']}")
print(f"\nMetadata:\n{json.dumps(result['metadata'], indent=2)}")
```
