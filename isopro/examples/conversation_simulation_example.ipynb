# Conversation Simulation Example using isopro

This notebook demonstrates how to use the conversation simulation with a Claude agent for customer service.

```python
import sys
import os

# Add the parent directory of the current working directory to the Python path
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
```

```python
from isopro.conversation_simulation import ConversationSimulator, ConversationAgent
from isopro.utils.conversation_analysis import analyze_conversation
import os
```

```python
# Set up the Claude agent with a customer service prompt
customer_service_prompt = """You are a helpful customer service representative for a tech company. 
Your goal is to assist customers with their inquiries politely and efficiently. 
Always maintain a professional tone and try to resolve issues in a timely manner."""

# Create the conversation simulator
simulator = ConversationSimulator(ai_prompt=customer_service_prompt)

# Run a simulation with an upset customer
conversation_history = simulator.run_simulation(persona_type="upset", num_turns=5)

# Print the conversation
for turn in conversation_history:
    print(f"{turn['role'].capitalize()}: {turn['content']}\n")

# Analyze the conversation
analysis_results = analyze_conversation(conversation_history)

print("Conversation Analysis:")
print(f"Overall sentiment: {analysis_results['overall_sentiment']:.2f}")
print(f"Sentiment trend: {analysis_results['sentiment_trend']:.2f}")
print(f"Context adaptation score: {analysis_results['context_adaptation_score']:.2f}")
```
